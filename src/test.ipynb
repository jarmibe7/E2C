```python
import torch
import numpy as np
import time
```

```python
num_action_samples = 50
plan_horizon = 5
mu = torch.randn((plan_horizon, 2))
sigma = torch.ones((plan_horizon, 2)) * 0.5
```

```python
costs = torch.zeros(num_action_samples)
action_samples = torch.zeros((num_action_samples, plan_horizon, mu.shape[1]))
trajs = []

# Sample action sequences from current distribution
tic = time.time()
for k in range(num_action_samples):
    a_t = torch.normal(mu, sigma)
    # print(a_t)

    # Clip to action space bounds
    action_samples[k] = torch.clamp(
        a_t, 
        torch.as_tensor(-1.0, dtype=torch.float32), 
        torch.as_tensor(1.0, dtype=torch.float32)
    )
    # print(action_samples[k])
toc = time.time()
print(f"Time taken to sample {num_action_samples} action sequences: {toc - tic} seconds")
```
